# TAI Public Finance: a computational guide

Workers do not automatically share in the capital-market gains from automation. This project asks whether a government can use taxes, safe borrowing or saving, and risky public assets to move resources toward the states in which workers value them most.

The notebooks develop one answer: a public balance sheet can insure workers along the payoff directions that markets offer, but it cannot reach risks outside that payoff span. Taxes remain relevant because they redistribute income and alter investment, although they cannot reproduce an asset payoff at the instant of a shock.

## The idea before the machinery

An internationally traded claim is priced by the marginal value of consumption to the investors who trade it. A worker-oriented government may value the same payoff differently because the payoff changes workers' consumption through the public budget. The key object is therefore the ratio of the government's fiscal valuation kernel to the market stochastic discount factor,

$$
\Theta=\frac{m_G}{m_I}.
$$

If this ratio varies across automation states, correct market pricing does not exhaust the government's insurance motive. The first two notebooks make that statement precise. The later notebooks ask how the motive propagates through capital, tax adjustment, public wealth, and uncertainty over which post-automation economy arrives.

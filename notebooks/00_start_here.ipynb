## The central question

Automation can affect workers' resources differently from the consumption of the investors who price capital-market claims. The project asks how far a government's tax system and balance sheet can complete workers' missing intertemporal and state-contingent markets when taxation is distortionary and the available public payoffs span only part of the relevant risk.

The notebooks build the answer in layers. They begin with the valuation and payoff-span mechanism, introduce the transparent local approximation, move to the nonlinear dynamic problem, and then consider uncertainty over whether automation leads to a labour-using or labour-free economy.

## The reading route

The repository manifest records the intended sequence and whether each notebook is available or still planned. The next cell reads that manifest directly, so the rendered route and the repository's maintained index cannot drift apart silently.

```python
import json
from pathlib import Path
from IPython.display import Markdown, display

manifest_candidates = [
    Path("notebooks/manifest.json"),
    Path("manifest.json"),
]
manifest_path = next(path for path in manifest_candidates if path.exists())
manifest = json.loads(manifest_path.read_text(encoding="utf-8"))

lines = [
    "| Order | Notebook | Reader question | Status |",
    "|---:|---|---|---|",
]
for item in manifest["notebooks"]:
    lines.append(
        f"| {item['order']} | {item['title']} | {item['reader_question']} | {item['status']} |"
    )

display(Markdown("\n".join(lines)))
```

## How to interpret what you see

The notebooks will distinguish four kinds of object:

- **Analytical result:** follows from the stated model and assumptions.
- **Computed result:** reproduced from a pinned implementation and identified run or frozen output.
- **Illustration:** chosen to explain a mechanism; it is not a calibrated quantitative finding.
- **Provisional or diagnostic result:** useful for understanding the work, but not yet suitable for a paper-level quantitative claim.

A polished chart does not change an object's status. Each result-bearing notebook will identify its source repository and commit, model branch, computational specification, and evidence.

# What taxes, safe finance, and risky claims each buy

Safe borrowing or saving moves resources across dates. A risky claim moves resources across states, but only along its payoff vector. A capital tax redistributes factor income and changes investment. These margins interact, so substitution or complementarity cannot be inferred from isolated coefficients or from calculations run under different model closures.

A welfare ranking requires matched policy problems that begin from the same inherited state $(S_0,M_0)$. Those matched numerical solutions are not yet available.

## Why the instruments are not interchangeable

| Instrument | Direct effect | Main interaction |
|---|---|---|
| Safe public position | Transfers resources across dates at the safe rate | Finances consumption or gradual tax adjustment after a shock |
| Risky public claim | Transfers resources along a marketed payoff direction | Changes the fiscal cost of insuring worker consumption in states reached by that payoff |
| Capital tax and tax speed | Redistribute rental income and alter capital accumulation | Respond to exposure left outside the traded payoff span, at an efficiency cost |

The safe position has no cross-state payoff at a given instant. The risky claim cannot hedge an orthogonal shock. The tax can reach distributional flows that the asset does not, but it changes private accumulation. A matched solution is needed because changing one instrument changes the shadow value and use of the others.

## The Brownian comparison that would identify the interactions

Hold technology, shocks, ownership, world prices, the worker's discounted log-utility objective, commitment, constraints, and the inherited $(S_0,M_0)$ fixed. Then solve six continuation problems: frozen or adjustable tax policy crossed with no risky claim, one claim, or two claims that span both Brownian shocks.

The cell labels below denote value functions for those six otherwise matched problems.

```python
from IPython.display import Markdown, display

rows = [
    ("Frozen tax", "$V_{F0}$", "$V_{F1}$", "$V_{F2}$"),
    ("Adjustable tax", "$V_{A0}$", "$V_{A1}$", "$V_{A2}$"),
]
menu_names = ["Tax rule", "No risky claim", "One risky claim", "Two-claim span"]

table = [
    "| " + " | ".join(menu_names) + " |",
    "|---|---|---|---|",
]
for label, *cells in rows:
    table.append("| " + " | ".join([label, *cells]) + " |")

display(Markdown("\n".join(table)))
```

If every restricted choice remains feasible in the richer menu, set inclusion alone gives the weak orderings

$$
V_{A2}\ge V_{A1}\ge V_{A0},
\qquad
V_{Am}\ge V_{Fm}
\quad\text{for }m\in\{0,1,2\}.
$$

These inequalities are conditional on a common feasible inherited state and genuinely nested choice sets. They are weak: they neither show a strict welfare gain nor reveal whether tax flexibility and asset access are substitutes or complements. The cross-difference

$$
\bigl(V_{A1}-V_{A0}\bigr)
-
\bigl(V_{F1}-V_{F0}\bigr)
$$

would answer that interaction question for one-claim access, but only after all four values are solved under the same closure.

## What can be concluded now

The structural conclusion is already useful: taxes, safe finance, and risky claims act on different margins and can change one another's value in equilibrium. Current public computation does not quantify that interaction or select an optimal mix.

The fixed-policy branch would need a frozen public output bundle and independent review before supporting group-specific consumption-equivalent figures. The Brownian instrument comparison would additionally need all six values from one model, one feasible $(S_0,M_0)$, a common financing closure, and uncertainty large enough that numerical error cannot reverse the ranking. Worker and domestic-owner welfare must remain separate unless a social objective and weights are explicitly specified; foreign welfare is outside the fixed-policy exercise.

# Instruments and welfare

**Reader question.** When do taxes, safe positions, and risky claims substitute for or
complement one another, and what can current code say about welfare?

**Answer in brief.** The instruments act on different margins: safe finance reallocates
across time, risky claims reallocate along their payoff span, and taxes redistribute
factor income while changing accumulation incentives. A valid quantitative ranking
must solve matched policy problems from the same inherited augmented state
$(S_0,M_0)$. That six-cell Brownian comparison does not yet have an accepted numerical
bundle.

Existing fixed-capital code evaluates **pre-announced rules** and is useful for tracing
accounting and incidence. It does not choose optimal policy. This notebook therefore
finishes the comparison contract and evidence map without reporting an unsupported
cross-instrument welfare ranking.

## Two different exercises

The main Brownian design is a two-by-three matrix: frozen versus adjustable tax policy,
crossed with no risky claim, one claim, or a completed two-claim menu. Every cell must
hold primitives, shock law, ownership, world prices, commitment, worker objective, and
the same feasible $(S_0,M_0)$ fixed.

The fixed-capital OU exercise instead fixes aggregate capital, domestic-private
ownership, tax rates, the transfer rule, and a world-SDF pricing closure. It asks what a
specified saving and portfolio rule does along a specified transition. It is not the
Brownian Ramsey economy with re-optimization “turned off,” because several other model
blocks also change.

## The matched Brownian comparison that is still required

The next cell renders the six-cell design and its current numerical status. Weak
feasible-set inequalities are available analytically when zero holdings and frozen tax
adjustment are feasible subsets. Strict welfare gains, policy signs, and magnitudes
require separate Bellman solutions in every cell.

```python
rows = [
    ("Frozen tax", "F0", "F1", "F2"),
    ("Adjustable tax", "A0", "A1", "A2"),
]
menu_names = ["Tax regime", "No risky claim", "One claim", "Completed two-claim span"]

table = [
    "| " + " | ".join(menu_names) + " |",
    "|---|---|---|---|",
]
for label, *cells in rows:
    rendered_cells = [f"**{cell}** — not numerically accepted" for cell in cells]
    table.append("| " + " | ".join([label, *rendered_cells]) + " |")

table.extend(
    [
        "",
        "Conditional feasible-set ordering: $V_{a2}\\ge V_{a1}\\ge V_{a0}$ and "
        "$V_{Am}\\ge V_{Fm}$ when the restricted choice is feasible from the same state.",
    ]
)
display(Markdown("\n".join(table)))
```

## Instrument map

| Instrument or margin | Direct channel | Interaction that creates complementarity | Comparison discipline |
|---|---|---|---|
| Safe public position | Shifts resources across dates | Can finance gradual tax adjustment after a shock | Same initial public net worth and safe-rate closure |
| Risky public claim | Transfers resources along a payoff vector | Changes the fiscal cost of maintaining worker consumption and tax paths | Same payoff process, menu, and inherited holding convention |
| Tax rate and tax speed | Redistribute rental income and affect capital accumulation | Can respond to exposure left outside the traded payoff span | Same adjustment cost, bounds, and commitment regime |

Substitution or complementarity is therefore an equilibrium property of a matched
model solution, not something identified by comparing the sizes of coefficients from
different repositories.

## What the fixed-policy implementation can eventually display

Within the fixed-capital OU model, a clean comparison holds the fund case, domestic-
equity share rule $\omega_D$, taxes, transfer rule, world SDF, and physical shock paths
fixed while changing the announced automation transition. A separate portfolio
contrast compares $\omega_D>0$ with safe-only $\omega_D=0$ within the same environment.

Worker and domestic-owner consumption equivalents are group-specific objects and must
not be added into a synthetic “social welfare” number. Foreign welfare is excluded.
Even a numerically small Bellman residual would not repair missing provenance, a changed
closure, or the absence of independent review.

## Publication gates

Before numerical welfare panels are added, the fixed-policy branch needs a frozen
public output bundle with hashes, exact inputs, environment lock, and independent
review. Before any tax-versus-portfolio ranking is added, the Brownian six-cell matrix
needs matched solutions from one feasible $(S_0,M_0)$, common primitives and financing
closure, uncertainty tables, and accepted finite- or small-risk status.

Until then, the defensible conclusion is structural: the instruments work on distinct
margins and can interact, but current accepted evidence does not quantify that
interaction or choose an optimal mix.
